```python
import pandas as pd
import sqlite3
```

```python
df = pd.read_csv("users.csv")
print("Data from CSV file:")
df
```

```python
conn = sqlite3.connect("users.db")
cursor = conn.cursor()
print("Database connected!")
```

```python
cursor.execute("DROP TABLE IF EXISTS users")

cursor.execute("""
CREATE TABLE users (
    id INTEGER PRIMARY KEY AUTOINCREMENT,
    name TEXT,
    email TEXT,
    location TEXT,
    phone TEXT
)
""")

conn.commit()
print("Table created!")
```

```python
for index, row in df.iterrows():
    cursor.execute(
        "INSERT INTO users (name, email, location, phone) VALUES (?, ?, ?, ?)",
        (row["name"], row["email"], row["location"], str(row["phone"]))
    )

conn.commit()
print(len(df), "records inserted successfully!")
```

```python
db_data = pd.read_sql_query("SELECT * FROM users", conn)
print("Data from SQLite database:")
db_data
```

```python
conn.close()
```
